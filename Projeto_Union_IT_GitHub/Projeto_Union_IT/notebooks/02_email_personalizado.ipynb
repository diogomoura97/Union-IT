# Etapa 2.2 — E-mail marketing com IA Generativa

Vamos usar um **modelo de linguagem já treinado**, por meio da API da OpenAI, para criar rascunhos de e-mails conforme o setor do lead.

**Importante:** esta etapa exige uma chave da API OpenAI, com acesso habilitado, e pode gerar custos de uso. Não publique sua chave no GitHub.

```python
import os
from getpass import getpass
from openai import OpenAI
```

## 1. Configurar a chave

Se você já configurou a variável de ambiente `OPENAI_API_KEY` no computador, pode pular esta célula. Senão, execute a célula abaixo: a chave será solicitada sem aparecer na tela. Se deixar vazio, o notebook mostrará somente a preparação dos textos, sem chamar a IA.

```python
# A chave fica apenas nesta sessão, sem ser escrita no notebook.
# Para informar a chave, remova o # da linha abaixo e execute esta célula.
# os.environ["OPENAI_API_KEY"] = getpass("Cole sua chave OpenAI: ")

print("Chave configurada?", bool(os.getenv("OPENAI_API_KEY")))
```

## 2. Criar instruções diferentes para cada setor

O setor **Tecnologia** pede um texto mais direto. Para **Saúde**, o tom é mais cuidadoso. A IA não deve inventar serviços ou promessas da Union IT.

```python
def preparar_pedido(setor):
    # O setor muda o tom e os assuntos do e-mail.
    orientacoes = {
        "Tecnologia": "Tom direto e moderno. Mencione desafios de processos digitais e agilidade.",
        "Saúde": "Tom cuidadoso e profissional. Mencione organização de processos e confiança.",
    }
    estilo = orientacoes.get(setor, "Tom profissional e cordial.")

    return f"""Escreva um rascunho de e-mail marketing B2B em português para um lead do setor de {setor}.
A Union IT está lançando um serviço digital de assinatura para empresas.
{estilo}
Inclua um assunto, uma saudação, um texto curto e um convite para conversar.
Não invente preço, funcionalidades, integrações, números, certificações ou resultados.
Não mencione dados pessoais do destinatário."""

print("Pedido para Tecnologia:\n")
print(preparar_pedido("Tecnologia"))
print("\nPedido para Saúde:\n")
print(preparar_pedido("Saúde"))
```

## 3. Pedir à IA para gerar o e-mail

Se a chave estiver configurada, o código envia as instruções ao modelo e imprime os dois rascunhos. Se não estiver, esta etapa é pulada sem erro.

```python
def gerar_email(setor):
    cliente = OpenAI()
    resposta = cliente.responses.create(
        model="gpt-4o-mini",
        instructions="Você escreve e-mails curtos, claros e profissionais para empresas brasileiras.",
        input=preparar_pedido(setor),
        max_output_tokens=350,
    )
    return resposta.output_text

if os.getenv("OPENAI_API_KEY"):
    for setor in ["Tecnologia", "Saúde"]:
        print(f"\n--- E-mail para {setor} ---\n")
        print(gerar_email(setor))
else:
    print("Geração não executada: configure OPENAI_API_KEY na etapa 1.")
```

A ferramenta cria rascunhos com linguagem específica para cada tipo de empresa. **Antes de enviar**, alguém do marketing deve revisar o texto.
